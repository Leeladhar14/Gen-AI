```python
from langchain_community.tools.requests.tool import RequestsGetTool
from langchain_core.tools import tool
from bs4 import BeautifulSoup
```

```python
requests_tool = RequestsGetTool()
requests_tool._config = {"allow_dangerous_requests": True}
```

```python
@tool
def extract_class_link(html: str) -> str:
    soup = BeautifulSoup(html, "html.parser")
    link = soup.find("a", string=lambda t: "class" in t.())
    return link["href"] if link else "No class link found."lower
```
