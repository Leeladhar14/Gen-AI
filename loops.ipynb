```python
from langchain.chat_models import init_chat_model
from typing_extensions import TypedDict
from typing import Literal, Optional
from langgraph.graph import StateGraph, START, END
```

```python
class GuessState(TypedDict):
    guess: int | None
    answer: int | None
    count: int | None
    min: int
    max: int 
    result: Literal["Win","Loose"]
```

```python
#  Node to count
from random import randint
def pick(state: GuessState) -> GuessState:
    #  pick the number and increase the count
    state['answer'] = randint(1,10)
    if state['count'] == None:
        state['count'] = 1
    else:
       state['count'] += 1 
    return state['count']
```

```python
def announce(state: GuessState) -> Literal["Win","Loose"]:
    # Announce the winner
    if state['guess'] == state['answer']:
        return "Win"
    else:
        return "Loose"
```

```python
def should_continue(state: StateGraph) -> Literal["exit", "continue"]:
    if state['guess'] == state['answer'] or state['count'] == 5:
        return "exit"
    else:
        return "continue"
```

```python
graph_builder = StateGraph(GuessState)
graph_builder.add_node("pick_number",pick)
graph_builder.add_node("results",announce)
graph_builder.add_edge(START,"pick_number")
graph_builder.add_conditional_edges(
    "pickanswer",
    should_continue,
    {
        "exit": "results",
        "continue": "pickanswer"
    }
)
graph_builder.add_edge("results",END)
guess_graph = graph_builder.compile()
```
